### CALCULO DE DISTANCIA CON SENSOR ULTRASONICO
---

La formula general de la velocidad del sonido:

$\boxed{
v=\frac{e}{t}
}$

Donde: 

$
e = \text{espacio total}
$

$
v = \text{velocidad del sonido en el medio}
$

$
t = \text{tiempo de la onda ultrasónica}
$

$\text{---------------------------------------------------------------------------}$

Ahora si tenemos un obstaculo frente al sensor, la onda del ultrasonico 

rebota en el obstaculo y vuelve al sensor



Asi la onda la distancia del ultrasonida que viaja la onda del 

ultrasonico es:

$
2d =e= v\,t
$

$
\boxed{
d=
\frac{v\,t}{2}
}
$

donde:

$
d = \text{distancia al objeto}
$

$
v = \text{velocidad del sonido en el medio}[m/s]
$

$
t = \text{tiempo de ida y retorno de la onda ultrasónica}[cm/\mu s]
$
