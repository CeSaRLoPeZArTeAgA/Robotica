```python
import numpy as np
from rot3D import Th,Rzh
np.set_printoptions(precision=4,suppress=True)
# construimos la matriz de transformación homogénea primero rotamos alrededor 
# del eje z 45 grados y luego trasladamos 2 unidades en x y 3 unidades en y
# H = T @ Rz
H = Th(2,3,0) @ Rzh(45)
print("H = \n",H)
```

```python
import numpy as np
from rot3D import Th,Rzh,Rxh,Ryh
np.set_printoptions(precision=4,suppress=True)
# construimos la matriz de transformación homogénea primero rotamos alrededor   
H=Th(2,3,4)@Ryh(53)@Rxh(60)@Rzh(45)
print("H = \n",H)
```

```python
import numpy as np
from rot3D import Th,Rzh
np.set_printoptions(precision=4,suppress=True)

H = Th(1,2,0) @ Rzh(45)
r_uvw = np.array([2,4,6,1])
r_xyz = H @ r_uvw
r_xyz = r_xyz[:3]

print("H = \n",H)
print("r_xyz = ",r_xyz)
```

```python
import numpy as np
from rot3D import Th,Rzh,Rxh,Ryh
np.set_printoptions(precision=4,suppress=True)

H = Th(2,3,4)@ Ryh(53) @ Rxh(60)@ Rzh(45)
r_uvw = np.array([2,4,6,1])
r_xyz = H @ r_uvw
#r_xyz = r_xyz[:3]

print("H = \n",H)
print("r_xyz = ",r_xyz[:3])

print("===== Recuperacion de las variable uvw ======")
r_uvw=np.linalg.inv(H) @ r_xyz
print(r_uvw[:3])
```
