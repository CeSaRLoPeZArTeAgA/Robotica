```python
import numpy as np
from homo4 import T4,Rz4
np.set_printoptions(precision=4,suppress=True)
# construimos la matriz de transformación homogénea primero rotamos alrededor 
# del eje z 45 grados y luego trasladamos 2 unidades en x y 3 unidades en y
# H = T @ Rz
H = T4(2,3,0) @ Rz4(45)
print("H = \n",H)
```

```python
import numpy as np
from homo4 import T4,Rz4,Rx4,Ry4
np.set_printoptions(precision=4,suppress=True)
H=T4(2,3,4)@Ry4(53)@Rx4(60)@Rz4(45)
print("H = \n",H)
```

```python
import numpy as np
from homo4 import T4,Rz4
np.set_printoptions(precision=4,suppress=True)

H = T4(1,2,0) @ Rz4(45)
r_uvw = np.array([2,4,6,1])
r_xyz = H @ r_uvw
r_xyz = r_xyz[:3]

print("H = \n",H)
print("r_xyz = ",r_xyz)
```

```python
import numpy as np
from homo4 import T4,Rz4,Rx4,Ry4
np.set_printoptions(precision=4,suppress=True)

H = T4(2,3,4)@ Ry4(53) @ Rx4(60)@ Rz4(45)
r_uvw = np.array([2,4,6,1])
r_xyz = H @ r_uvw
#r_xyz = r_xyz[:3]

print("H = \n",H)
print("r_xyz = ",r_xyz[:3])

print("===== Recuperacion de las variable uvw ======")
r_uvw=np.linalg.inv(H) @ r_xyz
print(r_uvw[:3])
```
