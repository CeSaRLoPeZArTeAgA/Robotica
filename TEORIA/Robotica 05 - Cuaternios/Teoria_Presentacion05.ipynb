```python
import numpy as np
from fquad import *
np.set_printoptions(precision=4,suppress=True)
 
u = np.array([3,-2,1])
k = u/np.linalg.norm(u)
Q=Rot(u,90)

print("Vector u =",u)
print("Vector unitario k =",k)
print("Matriz de rotacion alrededor de k equivalente cuaternio de rotacion Q")
print("Q =",Q)
```

```python
import numpy as np
from fquad import *
np.set_printoptions(precision=4,suppress=True)
 
r=np.array([5,2,-6])
u = np.array([3,-2,1])
k = u/np.linalg.norm(u)
Q=Rot(u,90)

print("Vector u =",u)
print("Vector unitario k =",k)
print("Matriz de rotacion alrededor de k equivalente cuaternio de rotacion Q")
print("Q =",Q)
print("Vector r =",r)
#ampliamos el cuaternio r a (0, r)
r = np.insert(r, 0, 0)
print("Vector rotado r' =",qprod(qprod(Q,r),qconj(Q)))
```

```python
import numpy as np
from fquad import *
np.set_printoptions(precision=4,suppress=True)
H=np.array([[1,0,0],[0,1/np.sqrt(2),-1/np.sqrt(2)],[0, 1/np.sqrt(2),1/np.sqrt(2)]])

print("Matriz de rotacion H = \n",H)
print("Cuaternio asociado q = ",h2q(H))
```

```python
import numpy as np
from fquad import *
np.set_printoptions(precision=6,suppress=True)

# pregunta a
k1 = np.array([1,0,1])
Q1=Rot(k1,45)
print("Respuesta a la pregunta a")
print("Q1=",Q1)

# pregunta b
k2 = np.array([1,1,0])
Q2=Rot(k2,30)
print("\nRespuesta a la pregunta b")
print("Q2=",Q2)

# pregunta c
p=np.array([1,1,1])
Q3=qprod(Q2,Q1)
print("\nRespuesta a la pregunta c")
print("Q3=",Q3)

#ampliamos el cuaternio p a (0, p)
p= np.insert(p, 0, 0)
print("Usando cuaternios: r'=",qprod(qprod(Q3,p),qconj(Q3)))

print("Matriz Homogenea H=")
print(q2h(Q3))

```
