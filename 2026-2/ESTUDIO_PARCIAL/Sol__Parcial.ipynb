```python
import numpy as np
from fquad import *
np.set_printoptions(precision=4,suppress=True)

v = np.array([1,-2,4]) 

a = np.array([1,2,2])
Q1=Rot(a,90)

b = np.array([3,0,0])
Q2=Rot(b,90)

Q3=qprod(Q2,Q1)
#ampliamos el cuaternio v a (0, v)
v= np.insert(v, 0, 0)

#impresion de resultados
print("Cuaternio Q1 de rotacion de 90 alrededor de a")
print("Q1 =",Q1)
print("\nCuaternio Q2 de rotacion de 90 alrededor de b")
print("Q2 =",Q2)
print("\nCuaternio de rotacion total equivalente Q = Q2*Q1")
print("Q3 =",Q3)
print("\nVector v =",v)
print("Usando cuaternios: w=v'=",qprod(qprod(Q3,v),qconj(Q3)))
```
