```python
import numpy as np
from fquad import *
np.set_printoptions(precision=4,suppress=True)

H1=T4(6,4,4)
H2=Rz4(-90)
H3=T4(-2,9,7)
H4=T4(2,3,-4)
H5=Ry4(-90)
H6=Rz4(-180)
H7=T4(7,0,2)
H8=Rx4(90)
H9=Rz4(-180)

H=H1@H2@H3@H4@H5@H6@H7@H8@H9

print("Matriz homogenea")
print(H)
```

```python
import numpy as np
from fquad import *
np.set_printoptions(precision=4,suppress=True)

# pregunta I
print("######   PREGUNTA I    ######\n")
q1=np.array([np.sqrt(3)/2,0.5,0,0]) 
print("cuaternio q1:",q1)
H1=q2h(q1)
print("Matriz homogena H1=")
print(H1)

q2=np.array([np.sqrt(2)/2,0,np.sqrt(2)/2,0])
print("\ncuaternio q2:",q2)
print("Matriz homogena H2=")
H2=q2h(q2)
print(H2)

q3=np.array([1/2,0,0,np.sqrt(3)/2]) 
print("\ncuaternio q3:",q3)
print("Matriz homogena H3=")
H3=q2h(q3)
print(H3)

qi=np.array([1,0,0,0]) 
print("\ncuaternio q_ini:",qi)
print("Matriz homogena Hini=")
Hi=q2h(qi)
print(Hi)

#punto inicial antes de rotaciones
p=np.array([2,1,3]) 

#matriz homogenea
H=H3@H2@H1@Hi
print("\nMatriz homogenea final:")
print(H)

#ampliamos el cuaternio p a (p,1)
ph = np.append(p, 1)

# aplicar la rotacion
pp = H @ ph

print("\nPunto inicial v=", p)
print("Punto rotado:", pp[:3])

# cuaternion final
qfin = qprod(q3, qprod(q2, q1))

print("\nCuaternion final:")
print(qfin)

# pregunta II
print("######   PREGUNTA II    ######\n")
qo = np.array([
    np.sqrt(3)/2,
    -1/np.sqrt(8),
    np.sqrt(3)/2,
    1/np.sqrt(8)
])

# normalizar el cuaternion objetivo
qo = qo / np.linalg.norm(qo)

# producto escalar
dot = np.dot(qfin, qo)

# error angular
theta = 2 * np.arccos(
    np.clip(np.abs(dot), 0, 1)
)

theta_deg = np.degrees(theta)

print("\nError angular (radianes):", theta)
print("Error angular (grados):", theta_deg)
```
